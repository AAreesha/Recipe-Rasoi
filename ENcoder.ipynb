```python
import pandas as pd
from sentence_transformers import CrossEncoder

# Load recipe data from Excel
file_path = "groundfinal.xlsx"  # Update the file name if needed
recipe_df = pd.read_excel(file_path)

# Combine recipe details into a single string for each recipe
recipe_df['combined_text'] = (
    recipe_df['Ingredients'] + " | " +
    recipe_df['Cuisine Region'] + " | " +
    recipe_df['Cooking Method'] + " | " +
    recipe_df['Recipe']
)

# Convert the combined text to a list
recipes = recipe_df['combined_text'].tolist()
```

```python
# Initialize a Cross-Encoder model
cross_encoder = CrossEncoder('cross-encoder/ms-marco-MiniLM-L-6-v2')
```

```python
# Get user input
user_input = input("Enter the ingredients you have (comma-separated): ").strip()
# Create pairs of (user_input, recipe) for all recipes
recipe_pairs = [(user_input, recipe) for recipe in recipes]
# Compute scores for each pair
scores = cross_encoder.predict(recipe_pairs)
```

```python
import numpy as np

# Sort indices by scores in descending order
top_n = 5  # Number of top recipes to display
top_indices = np.argsort(scores)[-top_n:][::-1]

print("\nTop Recipe Matches:")
for idx in top_indices:
    recipe_info = recipe_df.iloc[idx]
    print(f"Recipe Name: {recipe_info['Recipe Name']} (Score: {scores[idx]:.2f})")
    print(f"  Ingredients: {recipe_info['Ingredients']}")
    print(f"  Cuisine Region: {recipe_info['Cuisine Region']}")
    print(f"  Cooking Method: {recipe_info['Cooking Method']}")
    print(f"  Recipe: {recipe_info['Recipe']}")
    print(f"  Category: {recipe_info['Category']}\n")
```
