```python
import pandas as pd

# Load recipe data from Excel
file_path = "groundtruthfinal.xlsx" # Update the file name if needed
recipe_df = pd.read_excel(file_path)

# Replace 'Coastal Sindh' with 'Sindh' in the 'Cuisine Region' column
recipe_df['Cuisine Region'] = recipe_df['Cuisine Region'].replace('Punjabi', 'Punjab')
recipe_df['Cuisine Region'] = recipe_df['Cuisine Region'].replace('Mumbai', 'Fusion')
recipe_df['Cuisine Region'] = recipe_df['Cuisine Region'].replace('North Indian', 'Fusion')
recipe_df['Cuisine Region'] = recipe_df['Cuisine Region'].replace('Multan', 'Fusion')
recipe_df['Cuisine Region'] = recipe_df['Cuisine Region'].replace('South Indian', 'Fusion')
recipe_df['Cuisine Region'] = recipe_df['Cuisine Region'].replace('South India', 'Fusion')
recipe_df['Cuisine Region'] = recipe_df['Cuisine Region'].replace('Nationwide', 'Fusion')
recipe_df['Cuisine Region'] = recipe_df['Cuisine Region'].replace('Pakistani', 'Fusion')
recipe_df['Cuisine Region'] = recipe_df['Cuisine Region'].replace('Maharashtra', 'Fusion')
recipe_df['Cuisine Region'] = recipe_df['Cuisine Region'].replace('Mughlai', 'Fusion')
recipe_df['Cuisine Region'] = recipe_df['Cuisine Region'].replace('BBQ', 'Punjab')
recipe_df['Cuisine Region'] = recipe_df['Cuisine Region'].replace('Coastal regions', 'Coastal Region')

# Save the updated DataFrame to a new file
output_file = "groundfinal.xlsx"
recipe_df.to_excel(output_file, index=False)  # Save without the index column
```
