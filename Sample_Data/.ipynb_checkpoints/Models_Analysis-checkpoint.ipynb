```python
import pandas as pd

file_path = "kaggle_recipes.csv"
recipe_df = pd.read_csv(file_path)

recipe_df.head()
```

```python
import numpy as np
import pandas as pd
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline

# # Preprocess Ingredients
# vectorizer = TfidfVectorizer()
# X_ingredients = vectorizer.fit_transform(recipe_df['ingredients_list'])
```

```python
# Count occurrences of each recipe name
class_counts = recipe_df['recipe_name'].value_counts()

# Filter out classes that occur only once
valid_classes = class_counts[class_counts > 1].index
filtered_df = recipe_df[recipe_df['recipe_name'].isin(valid_classes)]
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split



X = filtered_df['ingredients_list']
y = filtered_df['recipe_name']

# Update test size based on the number of classes
n_classes = len(y.unique())
test_size = min(234, len(filtered_df) // 5)  # Ensure we have enough data for classes


print("X:",X)

print("Y:",y)
# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.5,      # 20% of the data for testing
    random_state=42,    # Seed for reproducibility
    stratify=y          # Maintain the distribution of recipe names
)

# Vectorize the text data
vectorizer = TfidfVectorizer()
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

# Check the sizes of the resulting sets
print(f'Training set size: {X_train.shape[0]}')
print(f'Testing set size: {X_test.shape[0]}')
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score
```

```python
# Define models to evaluate
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000),
    'Decision Tree': DecisionTreeClassifier(),
    'Random Forest': RandomForestClassifier(),
    'Support Vector Machine': SVC(),
    'Gradient Boosting': GradientBoostingClassifier()
}

# Train and evaluate each model
for model_name, model in models.items():
    # Train the model
    model.fit(X_train_tfidf, y_train)
    
    # Make predictions
    y_pred = model.predict(X_test_tfidf)
    
    # Evaluate the model
    accuracy = accuracy_score(y_test, y_pred)
    report = classification_report(y_test, y_pred)
    
    print(f'--- {model_name} ---')
    print(f'Accuracy: {accuracy:.2f}')
    # print('Classification Report:')
    # print(report)
    print('-----------------------------------')
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, VotingClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

# Assuming X_train_tfidf and y_train are already defined
# Define models to evaluate
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000),
    'Decision Tree': DecisionTreeClassifier(),
    'Random Forest': RandomForestClassifier(),
    'Support Vector Machine': SVC(),
    'Gradient Boosting': GradientBoostingClassifier(),
    'XGBoost': XGBClassifier(use_label_encoder=False, eval_metric='mlogloss'),
    'LightGBM': LGBMClassifier(),
    'CatBoost': CatBoostClassifier(silent=True),
    'Voting Classifier': VotingClassifier(estimators=[
        ('rf', RandomForestClassifier()),
        ('lr', LogisticRegression(max_iter=1000)),
        ('xgb', XGBClassifier(use_label_encoder=False, eval_metric='mlogloss'))
    ], voting='hard')
}

# Train and evaluate each model
for model_name, model in models.items():
    # Train the model
    model.fit(X_train_tfidf, y_train)
    
    # Make predictions
    y_pred = model.predict(X_test_tfidf)
    
    # Evaluate the model
    accuracy = accuracy_score(y_test, y_pred)
    
    print(f'--- {model_name} ---')
    print(f'Accuracy: {accuracy:.2f}')
    print('-----------------------------------')
```

```python
from sklearn.naive_bayes import MultinomialNB
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, SimpleRNN
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score
import numpy as np

# Encode the target labels for ANN and RNN
le = LabelEncoder()
y_train_enc = le.fit_transform(y_train)
y_test_enc = le.transform(y_test)

# Define a simple ANN model
def create_ann_model(input_dim):
    model = Sequential()
    model.add(Dense(64, input_dim=input_dim, activation='relu'))
    model.add(Dense(32, activation='relu'))
    model.add(Dense(len(np.unique(y_train_enc)), activation='softmax'))
    model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
    return model

# Define a simple RNN model
def create_rnn_model(input_dim):
    model = Sequential()
    model.add(SimpleRNN(50, input_shape=(input_dim, 1)))
    model.add(Dense(len(np.unique(y_train_enc)), activation='softmax'))
    model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
    return model

# Add Naive Bayes, ANN, and RNN to the models dictionary
models = {
    'Logistic Regression': LogisticRegression(max_iter=1000),
    'Decision Tree': DecisionTreeClassifier(),
    'Random Forest': RandomForestClassifier(),
    'Support Vector Machine': SVC(),
    'Gradient Boosting': GradientBoostingClassifier(),
    'Naive Bayes': MultinomialNB(),
    'Artificial Neural Network': create_ann_model(X_train_tfidf.shape[1]),  # Pass input size
    'Recurrent Neural Network': create_rnn_model(X_train_tfidf.shape[1])    # Pass input size
}

# Train and evaluate each model
for model_name, model in models.items():
    print(f'--- {model_name} ---')
    
    # Train Naive Bayes, Logistic Regression, Decision Tree, etc.
    if model_name in ['Naive Bayes', 'Logistic Regression', 'Decision Tree', 'Random Forest', 'Support Vector Machine', 'Gradient Boosting']:
        model.fit(X_train_tfidf, y_train)
        y_pred = model.predict(X_test_tfidf)
        accuracy = accuracy_score(y_test, y_pred)
    
    # Train ANN
    elif model_name == 'Artificial Neural Network':
        model.fit(X_train_tfidf.toarray(), y_train_enc, epochs=10, batch_size=32, verbose=0)
        y_pred = np.argmax(model.predict(X_test_tfidf.toarray()), axis=-1)
        accuracy = accuracy_score(y_test_enc, y_pred)
    
    # Train RNN
    elif model_name == 'Recurrent Neural Network':
        X_train_rnn = np.expand_dims(X_train_tfidf.toarray(), axis=-1)  # Reshape for RNN input
        X_test_rnn = np.expand_dims(X_test_tfidf.toarray(), axis=-1)
        model.fit(X_train_rnn, y_train_enc, epochs=10, batch_size=32, verbose=0)
        y_pred = np.argmax(model.predict(X_test_rnn), axis=-1)
        accuracy = accuracy_score(y_test_enc, y_pred)
    
    print(f'Accuracy: {accuracy:.2f}')
    print('-----------------------------------')
```
